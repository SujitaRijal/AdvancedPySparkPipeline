```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("Product Category Name Translation Profiling").getOrCreate()
```

```python
category_translation = spark.read.parquet("../data/bronze/translation")
```

```python
category_translation.printSchema()
```

```python
category_translation.count()
```

```python
category_translation.show(10,truncate=False)
```

```python
from pyspark.sql.functions import col,count,when
category_translation.select([
    count(when(col(c).isNull(),c)).alias(c)
    for c in category_translation.columns
]).show(vertical=True)
```

```python
duplicate_category_translation = category_translation.groupBy("product_category_name").count().filter(col("count")>1)
duplicate_category_translation.show()
print("Duplicate Product Category Translation:", duplicate_category_translation.count())
```
