```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName("Products Metrics Validation").getOrCreate()
```

```python
products_metrics= spark.read.parquet("../../data/gold/product_metrics")
```

```python
products_metrics.show(10, truncate=False)
```

```python
products_metrics.count()
```

```python
products_metrics.select("product_id").distinct().count()
```

```python
from pyspark.sql.functions import col
```

```python
products_metrics.orderBy(
    col("product_rank")
).show(10)
```

```python
products_metrics.filter(
    col("product_id").isNull()
).count()
```

```python
products_metrics.filter(
    col("total_sales") < 0
).count()
```

```python
products_metrics.filter(
    col("total_quantity_sold") <= 0
).count()
```
