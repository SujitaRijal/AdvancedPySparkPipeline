```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName("Customer Metrics Validation").getOrCreate()
```

```python
customer_metrics = spark.read.parquet("../../data/gold/customer_metrics")
```

```python
customer_metrics.show(10, truncate=False)
```

```python
from pyspark.sql.functions import col
```

```python
customer_metrics.groupBy("customer_unique_id") \
    .count() \
    .filter(col("count") > 1) \
    .show()
```
