```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("Sellers Profiling").getOrCreate()
```

```python
sellers = spark.read.parquet("../data/bronze/sellers")
```

```python
sellers.printSchema()
```

```python
sellers.count()
```

```python
from pyspark.sql.functions import col, when, count
sellers.select([
    count(when(col(c).isNull(),c)).alias(c)
    for c in sellers.columns
]).show(vertical=True)
```

```python
duplicate_sellers = sellers.groupBy("seller_id").count().filter(col("count") > 1)
duplicate_sellers.show()
print("Duplicate sellers:", duplicate_sellers.count())
```
