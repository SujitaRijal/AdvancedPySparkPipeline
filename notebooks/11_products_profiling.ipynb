```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("Silver Layer - Products").getOrCreate()
```

```python
products = spark.read.parquet("../data/bronze/products")
```

```python
products.printSchema()
```

```python
products.count()
```

```python
from pyspark.sql.functions import count, col, when
products.select([
    count(when(col(c).isNull(),c)).alias(c)
    for c in products.columns
]).show(vertical=True)
```

```python
duplicate_products = products.groupBy("product_id").count().filter(col("count")>1)
duplicate_products.show()
print("Duplicate products:", duplicate_products.count())
```
