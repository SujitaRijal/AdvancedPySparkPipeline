```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col
spark = SparkSession.builder.appName("Silver sellers validation").getOrCreate()
```

```python
bronze_sellers = spark.read.parquet("../data/bronze/sellers")
silver_sellers = spark.read.parquet("../data/silver/sellers")
```

```python
bronze_sellers.count()
```

```python
silver_sellers.count()
```

```python
from pyspark.sql.functions import col, when,count
silver_sellers.select([
    count(when(col(c).isNull(),c)).alias(c)
    for c in silver_sellers.columns
]).show(vertical=True)
```

```python
duplicate_silver_sellers = silver_sellers.groupBy("seller_id").count().filter(col("count")>1)
duplicate_silver_sellers.show()
print("Duplicate silver sellers", duplicate_silver_sellers.count())
```
