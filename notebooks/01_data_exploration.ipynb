```python
from pyspark.sql import SparkSession
```

```python
spark = (SparkSession.builder.appName("Data Exploration").getOrCreate())
```

```python
from pathlib import Path
```

```python
BASE_PATH = Path("../data/raw")
```

```python
customers = spark.read.option("header", True).csv(str(BASE_PATH / "olist_customers_dataset.csv"))
```

```python
orders = spark.read.option("header", True).csv(str(BASE_PATH / "olist_orders_dataset.csv"))
```

```python
order_items = spark.read.option("header", True).csv(str(BASE_PATH / "olist_order_items_dataset.csv"))
```

```python
payments = spark.read.option("header", True).csv(str(BASE_PATH / "olist_order_payments_dataset.csv"))
```

```python
products = spark.read.option("header", True).csv(str(BASE_PATH / "olist_products_dataset.csv"))
```

```python
sellers = spark.read.option("header", True).csv(str(BASE_PATH / "olist_sellers_dataset.csv"))
```

```python
reviews = spark.read.option("header", True).csv(str(BASE_PATH / "olist_order_reviews_dataset.csv"))
```

```python
translation = spark.read.option("header", True).csv(str(BASE_PATH / "product_category_name_translation.csv"))
```

```python
datasets = {
    "customers": customers,
    "orders": orders,
    "order_items" : order_items,
    "payments": payments,
    "products": products,
    "sellers": sellers,
    "reviews": reviews,
    "translation": translation
}

for name, df in datasets.items():
    print(f"{name}: {df.count()} rows, {len(df.columns)} columns")
```
